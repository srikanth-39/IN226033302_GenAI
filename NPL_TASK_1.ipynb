The word Love is capitalized while love is lowercase. As python is case sensitve, when we are performing tokenization it will be stored as different tokens. it will increase the vovabulary size unnecessarily.Converting everything to lowercase helps maintain consistency and improves model performance.

Q2.What happens if stopwords are not removed?
If stopwords are not removed, the dataset will contain many common words like "is", "the", "and", etc., which usually do not add much meaning.this can reduce the effeciency and accuracy of the model.It can also make the model slow.

Q3.Provide two real-world scenarios where removing stopwords can be harmful.
Sentiment Analysis: Words like "not" are very important. Removing them can completely change the meaning. Example: "not good" --> "good" (wrong meaning)

Question Answering Systems: Words like "who", "what", "where" are essential for understanding the question

Q4. Difference between stemming and lemmatization
Stemming cuts words to their root form using simple rules. Example: "running" --> "run", "studies" --> "studi" It may not always produce real words.

Lemmatization converts words to their correct base form using vocabulary and grammar rules. Example: "running" --> "run", "better" --> "good" It is more accurate but slower than stemming.

Task 2: Advanced Preprocessing Function

```python
import re

def preprocess_text(text):
    if not text or not isinstance(text, str):
        return []

    # Convert to lowercase
    text = text.lower()

    # Remove URLs and emails
    text = re.sub(r'http\S+|www\S+|https\S+|\S+@\S+', '', text)

    # Remove numbers
    text = re.sub(r'\d+', '', text)

    # Handle repeated characters
    text = re.sub(r'(.)\1{2,}', r'\1\1', text)

    # Remove special characters
    text = re.sub(r'[^a-z\s]', '', text)

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    tokens = text.split()

    # Remove very short tokens except 'no' and 'not'
    filtered_tokens = [
        word for word in tokens
        if len(word) > 2 or word in ["no", "not"]
    ]

    return filtered_tokens


# List of sentences
text = ["I have 2 dogs","This is   good","soooo goooood!!!","WOW!!! This is GREAT!!!","Visit http://example.com now"  ]

# Process each sentence
for sentence in text:
    result = preprocess_text(sentence)
    print("Original:", sentence)
    print("Tokens:", result)
    print("Cleaned:", " ".join(result))
    print("-" * 40)
```

Task 4: Token Analytics

```python
def token_analytics(tokens):
    if not tokens:
        return 0, 0, 0

    total_tokens = len(tokens)
    unique_tokens = len(set(tokens))
    avg_length = sum(len(word) for word in tokens) / total_tokens

    return total_tokens, unique_tokens, round(avg_length, 2)


for sentence in test_sentences:
    tokens = preprocess_text(sentence)
    total, unique, avg_len = token_analytics(tokens)

    print("Sentence:", sentence)
    print("Total Tokens:", total)
    print("Unique Tokens:", unique)
    print("Avg Length:", avg_len)
    print("-" * 50)
```

Task 6: Full Pipeline

```python
from collections import Counter

test_sentences = [
    "Get 100% FREE access now!!!",
    "I absolutely looooved this product 😍😍",
    "Worst service ever... 0/10",
    "Call me at 9876543210",
    "This is THE best course!!!",
    "Visit https://openai.com now!",
    "Nooooo this is baaad!!!",
    "OK OK OK I got it",
    "Win $$$ now!!! Limited offer!!!",
    "I am not happy with this"
]

def full_pipeline(text_list):
    all_tokens = []
    clean_sentences = []

    print("\n===== NLP Preprocessing Output =====\n")

    for text in text_list:
        tokens = preprocess_text(text)
        clean_sentence = " ".join(tokens)

        # Store results
        all_tokens.extend(tokens)
        clean_sentences.append(clean_sentence)

        # Print per sentence output
        print("Original:", text)
        print("Tokens:", tokens)
        print("Cleaned Sentence:", clean_sentence)
        print("-" * 50)

    # Frequency analysis
    counter = Counter(all_tokens)

    print("\n===== Frequency Analysis =====\n")
    print("Top 10 most frequent words:")
    print(counter.most_common(10))

    print("\nTop 5 least frequent words:")
    print(counter.most_common()[-5:])

    return {
        "tokens": all_tokens,
        "clean_sentences": clean_sentences
    }
# Run the pipeline
result = full_pipeline(test_sentences)
```
